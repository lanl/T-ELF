```python
from TELF.factorization import ShiftNMFkTFGM
import matplotlib.pyplot as plt
from matplotlib.patches import Ellipse
import pandas as pd
import numpy as np

def tdoa_frames_from_positions(P, Spos, c, hop_sec, ref=0):
    N_ = P.shape[0] 
    K_ = Spos.shape[0]
    tau = np.zeros((N_, K_))
    for i in range(K_):
        d = np.linalg.norm(P - Spos[i], axis=1)
        dt = (d - d[ref]) / c
        tau[:, i] = dt / hop_sec
    return tau

from TELF.factorization.DopplerShiftNMFk import reconstruct_from_params, \
    bootstrap_triangulation, crb_covariance, unwrap_bins, \
    add_noise_by_snr_clipped_per_sensor, reestimate_tau_f_shifts_only

# ---- Parameters & geometry
N, Fdim, Tdim, K = 3, 96, 64, 2
SENSOR_POS = np.array([[0.0, 0.0],[2.0, 0.0],[0.6, 1.8]], dtype=float)
TRUE_SRC_POS = np.array([[1.1, 0.9],[1.7, 0.3]], dtype=float)
HOP_SEC = 0.01
C_PROP = 343.0
BIN_HZ = 1.0

# ---- Synthetic data
rng = np.random.default_rng(0)
f = np.arange(Fdim)
t = np.arange(Tdim)
true_F = np.stack([np.exp(-0.5*((f-28)/5)**2), np.exp(-0.5*((f-62)/7)**2)])
true_S = np.stack([np.exp(-0.5*((t-18)/4)**2), np.exp(-0.5*((t-44)/6)**2)])
true_F = true_F / (true_F.sum(axis=1, keepdims=True) + 1e-12)
true_S = true_S / (true_S.sum(axis=1, keepdims=True) + 1e-12)
true_W = rng.random((N, K))
true_tau_f = rng.uniform(-1.5, 1.5, size=(N, K))


true_tau_t = tdoa_frames_from_positions(SENSOR_POS, TRUE_SRC_POS, C_PROP, HOP_SEC, ref=0)

X_clean = reconstruct_from_params(true_W, true_F, true_S, true_tau_f, true_tau_t)
SNR_DB = 30.0
X = add_noise_by_snr_clipped_per_sensor(X_clean, snr_db=SNR_DB, rng=rng)

print("Norms ||X[n]||:", [float(np.linalg.norm(X[n])) for n in range(N)])
print("Mean |tau_t| frames:", float(np.mean(np.abs(true_tau_t))))

nmfk_params = {
    "n_perturbs":12,
    "n_iters":100,
    "epsilon":0.01,
    "n_jobs":-1,
    "init":"nnsvd",
    "use_gpu":False,
    "save_path":"../../results/",
    "verbose":True,
    "sill_thresh":0.8,
    "nmf_method":"nmf_fro_mu", 
    "perturb_type":"uniform", 
    "calculate_error":True,
    "pruned":True,
    "predict_k":True,
    "predict_k_method":"sill",
    "transpose":False,
    "mask":None,
    "use_consensus_stopping":False,
    "calculate_pac":True,
    "consensus_mat":True,
    "simple_plot":True,
    "collect_output":True
}

# X is (N, F, T), nonnegative
Ks = [1, 2, 3, 4, 5]

model = ShiftNMFkTFGM(
    experiment_name="MyShiftNMFkTFGM",
    save_path="results_shift_tfgm",
    mode="staged",             # or "monolithic" / "shifts_only"
    n_inits=13,
    n_jobs=-1,
    verbose=True,
    seed_base=0,
    # v12 knobs:
    ridge_w=0.0,
    iters_w=60, iters_fs=120, iters_snap=80,
    snap_window_f=3, snap_window_t=3,
    inner_rank1_iters=1,
)

# (Optional) use classic NMFk on a 2D flattening to guide K
# nmfk_diag = model.fit_nmfk(X, Ks, note="flattened-NF-by-T")
# Inspect nmfk_diag, PAC curves, etc., to shortlist K if desired

# Run the arrays-only Shift-NMFk-TFGM across all Ks
allK = model.fit_shift_nmfk(X, Ks)

# Choose a K programmatically, e.g., min PAC or min validation loss
k_star = min(allK.keys(), key=lambda k: (allK[k]["pac"] if allK[k]["pac"] is not None else 1.0, allK[k]["loss"]))
best = allK[k_star]
print("K* =", k_star, "loss=", best["loss"], "relerr=", best["relerr"])
print("Saved file:", best["save_file"])
```

```python
# ONE BLOCK: stable indexing + export artifacts with optional alignment ("truth" or "seed")

from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Ellipse
from scipy.optimize import linear_sum_assignment

# ---------- helpers: geometry / ellipses ----------
def _cov_ellipse(mu, cov, nsig=2.0, **kw):
    w, v = np.linalg.eigh(cov)
    w = np.clip(w, 0, None)
    order = np.argsort(w)[::-1]
    w, v = w[order], v[:, order]
    width, height = 2*nsig*np.sqrt(w)
    angle = np.degrees(np.arctan2(v[1,0], v[0,0]))
    return Ellipse(xy=mu, width=width, height=height, angle=angle, fill=False, **kw)

def _ellipse_params(cov, nsig=2.0):
    w, v = np.linalg.eigh(cov)
    idx = np.argsort(w)[::-1]
    w, v = w[idx], v[:, idx]
    a = nsig*np.sqrt(max(w[0], 0.0))
    b = nsig*np.sqrt(max(w[1], 0.0))
    ang = np.degrees(np.arctan2(v[1,0], v[0,0]))
    area = np.pi*a*b
    return a, b, ang, area

# ---------- helpers: alignment (truth or seed) ----------
def align_components_by_truth(pos_hat, true_pos):
    """Permutation so that pos_hat_aligned[i] ~= true_pos[i]."""
    diff = pos_hat[:, None, :] - true_pos[None, :, :]
    D = np.linalg.norm(diff, axis=2)             # (K,K)
    row_ind, col_ind = linear_sum_assignment(D)  # est row -> true col
    perm = row_ind[np.argsort(col_ind)]          # learned indices in truth order
    return perm

def _best_circ_corr(a, b):
    """Shift-invariant normalized circular correlation (max over shifts)."""
    eps = 1e-12
    A = np.fft.fft(a); B = np.fft.fft(b)
    c = np.real(np.fft.ifft(np.conj(A)*B))
    c /= ((np.linalg.norm(a)+eps)*(np.linalg.norm(b)+eps))
    return float(np.max(c))

def permute_to_seed_order(F_learn, S_learn, F_seed, S_seed, wF=0.5, wS=0.5):
    """Permutation that maximizes similarity to seeds (shift-invariant)."""
    K = F_seed.shape[0]
    C = np.zeros((K, K))
    for i in range(K):
        for j in range(K):
            C[i, j] = wF*_best_circ_corr(F_learn[i], F_seed[j]) + wS*_best_circ_corr(S_learn[i], S_seed[j])
    row_ind, col_ind = linear_sum_assignment(-C)  # maximize similarity
    perm = row_ind[np.argsort(col_ind)]
    return perm

def _apply_perm_everywhere(perm, W, F, S, tau_f, tau_t, pos_hat=None, covs=None, dt_base=None, dt_samps=None):
    W  = W[:, perm]
    F  = F[perm, :]
    S  = S[perm, :]
    tau_f = tau_f[:, perm]
    tau_t = tau_t[:, perm]
    if pos_hat is not None: pos_hat = pos_hat[perm]
    if covs    is not None: covs    = [covs[i] for i in perm]
    if dt_base is not None: dt_base = dt_base[:, perm]
    if dt_samps is not None: dt_samps = dt_samps[:, :, perm]
    return W, F, S, tau_f, tau_t, pos_hat, covs, dt_base, dt_samps

# ---------- MAIN: export artifacts with stable component indices ----------
def export_shift_tfgm_artifacts(
    allK: dict,
    k: int,
    X: np.ndarray,
    SENSOR_POS: np.ndarray,
    HOP_SEC: float,
    C_PROP: float,
    BIN_HZ: float,
    out_dir: Path,
    *,
    B_bootstrap: int = 200,
    SNR_DB: float = 30.0,
    rng_seed: int = 123,
    TRUE_SRC_POS: np.ndarray | None = None,   # (K,2) optional
    TRUE_TAU_F: np.ndarray | None = None,     # (N,K) optional
    BASE_X: np.ndarray | None = None,         # optional clean base
    show_plot: bool = True,
    filename_suffix_with_k: bool = True,
    align: str | None = None,                 # None | "truth" | "seed"
    F_seed: np.ndarray | None = None,         # needed if align="seed"
    S_seed: np.ndarray | None = None,         # needed if align="seed"
):
    """
    Recreates triangulation figure and CSVs with component indices stabilized by `align`.
    Returns dict of file paths.
    """
    from TELF.factorization.DopplerShiftNMFk import (
        reconstruct_from_params, bootstrap_triangulation, crb_covariance, unwrap_bins,
        add_noise_by_snr_clipped_per_sensor, reestimate_tau_f_shifts_only
    )

    out_dir = Path(out_dir); out_dir.mkdir(parents=True, exist_ok=True)
    res = allK[k]
    W, F, S, tau_f, tau_t = res["W"].copy(), res["F"].copy(), res["S"].copy(), res["tau_f"].copy(), res["tau_t"].copy()
    N, K = W.shape[0], F.shape[0]

    # Base (clean) cube for bootstrap
    base = BASE_X if BASE_X is not None else reconstruct_from_params(W, F, S, tau_f, tau_t)

    # Triangulation + bootstrap (before alignment so we can also permute bootstrap arrays)
    rng = np.random.default_rng(rng_seed)
    pos_hat, covs, tri_raw = bootstrap_triangulation(
        W, F, S, tau_f, tau_t, SENSOR_POS, HOP_SEC, C_PROP, ref=0,
        B=B_bootstrap, rng=rng, snr_db=SNR_DB, base_X=base
    )
    dt_base  = tri_raw["dt_base"]    # (N,K)
    dt_samps = tri_raw["dt_samples"] # (B,N,K)

    # ---- ALIGN INDICES (stable identity)
    if align == "truth":
        if TRUE_SRC_POS is None:
            raise ValueError("align='truth' requires TRUE_SRC_POS.")
        perm = align_components_by_truth(pos_hat, TRUE_SRC_POS)
        W, F, S, tau_f, tau_t, pos_hat, covs, dt_base, dt_samps = _apply_perm_everywhere(
            perm, W, F, S, tau_f, tau_t, pos_hat, covs, dt_base, dt_samps
        )
    elif align == "seed":
        if F_seed is None or S_seed is None:
            raise ValueError("align='seed' requires F_seed and S_seed.")
        perm = permute_to_seed_order(F, S, F_seed, S_seed, wF=0.5, wS=0.5)
        W, F, S, tau_f, tau_t, pos_hat, covs, dt_base, dt_samps = _apply_perm_everywhere(
            perm, W, F, S, tau_f, tau_t, pos_hat, covs, dt_base, dt_samps
        )
    # else: None -> keep solver order

    # CRB after (potential) reordering
    sigma_dt_est = dt_samps.std(axis=(0,2))  # (N,)
    covs_crb = [crb_covariance(SENSOR_POS, dt_base[:, i], C_PROP, ref=0,
                               sigma_tdoa_sec=sigma_dt_est, s_eval=pos_hat[i])
                for i in range(K)]

    # ---- Figure
    fig, ax = plt.subplots(figsize=(6,6))
    ax.scatter(SENSOR_POS[:,0], SENSOR_POS[:,1], marker='^', s=100, c="k", label="sensors", zorder=2)
    for i in range(N):
        ax.text(SENSOR_POS[i,0], SENSOR_POS[i,1], f" {i}", va='bottom', fontsize=9)
    if TRUE_SRC_POS is not None:
        ax.scatter(TRUE_SRC_POS[:,0], TRUE_SRC_POS[:,1], marker='*', s=160, c="tab:green", label="true src", zorder=3)
    ax.scatter(pos_hat[:,0], pos_hat[:,1], facecolors='none', edgecolors="tab:blue",
               marker='o', s=90, linewidths=2, label="est src", zorder=4)

    colors = ["tab:blue","tab:orange","tab:red","tab:purple","tab:brown","tab:cyan"]
    for i in range(K):
        e_boot = _cov_ellipse(pos_hat[i], covs[i], nsig=2.0,
                              edgecolor=colors[i%len(colors)], linewidth=2.5, zorder=5,
                              label="bootstrap 2σ" if i==0 else None)
        ax.add_patch(e_boot)
        if covs_crb[i] is not None:
            e_crb = _cov_ellipse(pos_hat[i], covs_crb[i], nsig=2.0,
                                 edgecolor=colors[i%len(colors)], linestyle='--', linewidth=2.0, zorder=5,
                                 label="CRB 2σ" if i==0 else None)
            ax.add_patch(e_crb)
        ax.text(pos_hat[i,0], pos_hat[i,1], f"  src {i}", color=colors[i%len(colors)], zorder=6)

    ax.set_aspect('equal', 'box'); ax.set_xlabel("x [m]"); ax.set_ylabel("y [m]")
    ax.set_title("Triangulation — bootstrap vs CRB (2σ)")
    ax.grid(True); ax.legend(loc="best")

    suf = f"_K={k}" if filename_suffix_with_k else ""
    fig_path = Path(out_dir) / f"triangulation_bootstrap_vs_crb{suf}.png"
    fig.savefig(fig_path, dpi=150, bbox_inches="tight")
    if show_plot: plt.show()
    else: plt.close(fig)

    # ---- CSV: positions / uncertainties
    rows = []
    for i in range(K):
        ab, bb, angb, areab = _ellipse_params(covs[i], nsig=2.0)
        if covs_crb[i] is not None:
            acrb, bcrb, angc, areac = _ellipse_params(covs_crb[i], nsig=2.0)
        else:
            acrb = bcrb = angc = areac = np.nan
        if TRUE_SRC_POS is not None:
            err = float(np.hypot(pos_hat[i,0]-TRUE_SRC_POS[i,0], pos_hat[i,1]-TRUE_SRC_POS[i,1]))
            tx, ty = TRUE_SRC_POS[i,0], TRUE_SRC_POS[i,1]
        else:
            err = np.nan; tx = ty = np.nan
        rows.append(dict(
            source=i, true_x=tx, true_y=ty, est_x=pos_hat[i,0], est_y=pos_hat[i,1],
            err_euclid=err,
            boot_semimajor_m=ab, boot_semiminor_m=bb, boot_angle_deg=angb, boot_area_m2=areab,
            crb_semimajor_m=acrb, crb_semiminor_m=bcrb, crb_angle_deg=angc, crb_area_m2=areac
        ))
    df_pos = pd.DataFrame(rows, columns=[
        "source","true_x","true_y","est_x","est_y","err_euclid",
        "boot_semimajor_m","boot_semiminor_m","boot_angle_deg","boot_area_m2",
        "crb_semimajor_m","crb_semiminor_m","crb_angle_deg","crb_area_m2"
    ])
    tri_csv = Path(out_dir) / f"triangulation_bootstrap_vs_crb{suf}.csv"
    df_pos.to_csv(tri_csv, index=False)

    # ---- Doppler CSVs
    Fbins = F.shape[1]
    from TELF.factorization.DopplerShiftNMFk import unwrap_bins, add_noise_by_snr_clipped_per_sensor, reestimate_tau_f_shifts_only
    fd_est = unwrap_bins(tau_f, Fbins) * float(BIN_HZ)      # (N, K)
    fd_true = unwrap_bins(TRUE_TAU_F, Fbins) * float(BIN_HZ) if TRUE_TAU_F is not None else None

    B = B_bootstrap
    fd_samps = np.zeros((B, N, K))
    for b in range(B):
        Xb = add_noise_by_snr_clipped_per_sensor(base, snr_db=SNR_DB, rng=np.random.default_rng(rng_seed + b))
        tauf_b = reestimate_tau_f_shifts_only(Xb, W, F, S, tau_f, tau_t)
        fd_samps[b] = unwrap_bins(tauf_b, Fbins) * float(BIN_HZ)
    fd_sigma = fd_samps.std(axis=0)

    rows = []
    for n in range(N):
        for i in range(K):
            rows.append(dict(
                sensor=n, source=i,
                fd_true=(fd_true[n,i] if fd_true is not None else np.nan),
                fd_est=fd_est[n,i], fd_sigma=fd_sigma[n,i]
            ))
    df_fd = pd.DataFrame(rows, columns=["sensor","source","fd_true","fd_est","fd_sigma"])
    doppler_csv = Path(out_dir) / f"doppler_frequency_per_sensor{suf}.csv"
    df_fd.to_csv(doppler_csv, index=False)

    rows_src = []
    for i in range(K):
        w = W[:, i].copy(); w = w / (w.sum() + 1e-12)
        fd_mean_est = float(np.sum(w * fd_est[:, i]))
        fd_mean_samples = np.sum(fd_samps[:, :, i] * w[None, :], axis=1)
        fd_mean_sigma = float(fd_mean_samples.std())
        rows_src.append(dict(
            source=i,
            fd_true_mean=(float(np.sum(w * fd_true[:, i])) if fd_true is not None else np.nan),
            fd_est_mean=fd_mean_est,
            fd_sigma_mean=fd_mean_sigma
        ))
    df_fd_src = pd.DataFrame(rows_src, columns=["source","fd_true_mean","fd_est_mean","fd_sigma_mean"])
    doppler_src_csv = Path(out_dir) / f"doppler_frequency_per_source{suf}.csv"
    df_fd_src.to_csv(doppler_src_csv, index=False)

    return {
        "triangulation_png": str(fig_path),
        "triangulation_csv": str(tri_csv),
        "doppler_per_sensor_csv": str(doppler_csv),
        "doppler_per_source_csv": str(doppler_src_csv),
    }
```

```python
# FIXED: align to SEED order for EVERY K (handles K != seed_K).
# - If K <= seed_K: uses the first K seeds (or the K best-matching seeds if you prefer).
# - If K > seed_K: falls back to a deterministic canonical sort (no seed match).
# This prevents IndexError: "index 1 is out of bounds ..." when K=1 but seed_K=2.

import numpy as np
from scipy.optimize import linear_sum_assignment

# --------- shift-invariant similarity ----------
def _best_circ_corr(a: np.ndarray, b: np.ndarray) -> float:
    eps = 1e-12
    A = np.fft.fft(a); B = np.fft.fft(b)
    c = np.real(np.fft.ifft(np.conj(A)*B))
    return float(np.max(c) / ((np.linalg.norm(a)+eps)*(np.linalg.norm(b)+eps)))

def _similarity_matrix_seed(F_learn: np.ndarray, S_learn: np.ndarray,
                            F_seed: np.ndarray,  S_seed: np.ndarray,
                            wF: float = 0.5, wS: float = 0.5) -> np.ndarray:
    KL = F_learn.shape[0]; KS = F_seed.shape[0]
    C = np.zeros((KL, KS), dtype=float)
    for i in range(KL):
        for j in range(KS):
            C[i, j] = wF*_best_circ_corr(F_learn[i], F_seed[j]) + wS*_best_circ_corr(S_learn[i], S_seed[j])
    return C

# --------- canonical (seed-free) stable order ----------
def _canonical_sort_perm(F: np.ndarray, S: np.ndarray, tau_t: np.ndarray | None = None) -> np.ndarray:
    f = np.arange(F.shape[1]); t = np.arange(S.shape[1])
    Fsum = F.sum(axis=1, keepdims=True) + 1e-12
    Ssum = S.sum(axis=1, keepdims=True) + 1e-12
    f_cent = (F @ f) / Fsum.ravel()
    t_cent = (S @ t) / Ssum.ravel()
    if tau_t is not None:
        tau_mean = tau_t.mean(axis=0)
        keys = np.stack([f_cent, t_cent, tau_mean], axis=1)
    else:
        keys = np.stack([f_cent, t_cent], axis=1)
    # sort by f_cent, then t_cent, then tau_mean
    return np.lexsort(keys[:, ::-1].T)

# --------- apply permutation everywhere ----------
def _apply_perm_res(res: dict, perm: np.ndarray) -> None:
    res["W"]     = res["W"][:, perm]
    res["F"]     = res["F"][perm, :]
    res["S"]     = res["S"][perm, :]
    res["tau_f"] = res["tau_f"][:, perm]
    res["tau_t"] = res["tau_t"][:, perm]
    if "pos_hat"   in res and res["pos_hat"]   is not None: res["pos_hat"]   = res["pos_hat"][perm]
    if "covs"      in res and res["covs"]      is not None: res["covs"]      = [res["covs"][i] for i in perm]
    if "dt_base"   in res and res["dt_base"]   is not None: res["dt_base"]   = res["dt_base"][:, perm]
    if "dt_samples" in res and res["dt_samples"] is not None: res["dt_samples"] = res["dt_samples"][:, :, perm]
    res["perm_applied"] = perm.copy()

# --------- public: align allK to seed order with fallbacks ----------
def align_allK_to_seed_inplace(allK: dict,
                               F_seed: np.ndarray,
                               S_seed: np.ndarray,
                               wF: float = 0.5,
                               wS: float = 0.5,
                               verbose: bool = True) -> dict:
    """
    Reorder components in allK so that indices are stable.
    If K <= seed_K: match to first K seeds (by similarity + Hungarian).
    If K > seed_K : fall back to canonical (seed-free) sort for that K.
    Returns {K: perm_used}.
    """
    seed_K = F_seed.shape[0]
    perms = {}
    for Kval, res in allK.items():
        KL = res["F"].shape[0]
        if KL == 1 and seed_K >= 1:
            # trivial, pick the single best-matching seed[0] (or just [0])
            perm = np.array([0], dtype=int)
            _apply_perm_res(res, perm)
            perms[Kval] = perm
            if verbose: print(f"[align] K={Kval}: single component -> perm {perm.tolist()}")
            continue

        if KL <= seed_K:
            # use the first KL seeds; if you want "best KL seeds", use Hungarian on full bank then
            F_seed_sub = F_seed[:KL, :]
            S_seed_sub = S_seed[:KL, :]
            C = _similarity_matrix_seed(res["F"], res["S"], F_seed_sub, S_seed_sub, wF=wF, wS=wS)
            row_ind, col_ind = linear_sum_assignment(-C)   # maximize similarity
            # learned row -> seed col; reorder learned to seed order:
            perm = row_ind[np.argsort(col_ind)]
            _apply_perm_res(res, perm)
            perms[Kval] = perm
            if verbose: print(f"[align] K={Kval}: seed match perm {perm.tolist()}")
        else:
            # more components than seeds -> canonical stable order
            perm = _canonical_sort_perm(res["F"], res["S"], res.get("tau_t"))
            _apply_perm_res(res, perm)
            perms[Kval] = perm
            if verbose: print(f"[align] K={Kval}: fallback canonical perm {perm.tolist()}")
    return perms
```
